```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# read in csv
df = pd.read_csv("2024-annual-report-report-data-all-states.csv", encoding='latin1', usecols=['Measure','State','Value','Rank','Source'])
df_depression_distribution = df[ 
                (df['Source'] == 'CDC, Behavioral Risk Factor Surveillance System') & 
                (df['Measure'] == 'Depression')]
df_depression_distribution = df_depression_distribution.drop( columns = ['Source'])
```

```python
# save data frame
df_depression_distribution.to_csv("Depression_ValueDistribution_AllStates.csv", index = False)
```
