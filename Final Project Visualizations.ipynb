```python
import csv
import pandas as pd
```

# HAWAII

```python
HIdf = pd.read_csv('data/5states_depression_value&score_&OUTSIDE_VARS.csv', usecols = ["Unnamed: 0", "HI"])

HIdf2 = HIdf.drop(0)
HIdf2 = HIdf2.rename(columns = {"Unnamed: 0": "Variables", "HI": "Value"})
HIdf2["Variables"] = HIdf2["Variables"].astype("string")
HIdf2["Value"] = HIdf2["Value"].astype("float")
```

```python
HIdf2
```

```python
import seaborn as sns
```

```python
colors = sns.color_palette("hls", 5)
HIplot = sns.barplot(
    x = "Variables",
    y = "Value",
    data = HIdf2,
    palette = colors
)

HIplot.set_title("Hawaii")
HIplot.set_ylim(0, 50) 
HIplot.tick_params(axis='x', rotation=25)

for label in HIplot.get_xticklabels():
    label.set_horizontalalignment('right')
```

```python
HIfig = HIplot.get_figure()
HIfig
```

```python
HIfig.savefig("figures/Hawaii_OtherVars.jpg", bbox_inches = 'tight')
```

# Depression Value x State Depression Rankings

```python
df = pd.DataFrame({
    "Depression State Rankings": [1, 13, 25, 36, 48],
    "Depression Value": [13.9, 19.2, 22.3, 24.1, 31.1],
    "Poverty Value": [10.3, 11.6, 11.1, 13.8, 17.2],
    "Food Insecurity Value": [9.6, 9.8, 8.6, 13, 13.7],
    "Adverse Childhood Experiences Value": [13.2, 16.3, 17.9, 18.2, 20],
    "Insufficient Sleep Value": [45.6, 32.9, 33.2, 35.2, 41.3],
    "label": ["Hawaii", "Iowa", "North Dakota", "Michigan", "West Virginia"]
})

DepressionByRankPlot = sns.regplot(
    x = "Depression State Rankings",
    y = "Depression Value",
    data = df,
    color = 'orange'
)

DepressionByRankPlot.set_title("Depression Value by State Depression Ranking")
for i, row in df.iterrows():
    DepressionByRankPlot.text(row["Depression State Rankings"], row["Depression Value"], row["label"])
```

```python
DepressionByRankfig = DepressionByRankPlot.get_figure()
DepressionByRankfig.savefig("figures/DepressionValueByRank.jpg", bbox_inches = 'tight')
```
